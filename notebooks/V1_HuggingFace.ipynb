````python
# Cell 6: Upload everything to Hugging Face
from huggingface_hub import HfApi, upload_file, upload_folder
import os

# Configuration
REPO_ID = "ranasinghehashini/srilankan-food-recognition"
SAVE_DIR = '/content/drive/MyDrive/Research-2026/SriLankanFoodRecognition'

api = HfApi()

print("="*70)
print("📤 UPLOADING TO HUGGING FACE")
print("="*70)
print(f"Repository: https://huggingface.co/{REPO_ID}\n")

# Files to upload
files = [
    ('best_model.pth', 'PyTorch model weights'),
    ('model_torchscript.pt', 'TorchScript model (mobile)'),
    ('model_info.json', 'Model metadata'),
    ('confusion_matrix.png', 'Confusion matrix'),
    ('per_class_accuracy.png', 'Per-class accuracy chart'),
    ('training_curves.png', 'Training curves'),
    ('classification_report.txt', 'Classification report'),
]

# Upload README first
print("1. Uploading Model Card...")
try:
    api.upload_file(
        path_or_fileobj='/content/README.md',
        path_in_repo='README.md',
        repo_id=REPO_ID,
        repo_type='model',
    )
    print("   ✅ README.md uploaded")
except Exception as e:
    print(f"   ❌ Error: {e}")

# Upload exported model folder
print("\n2. Uploading exported model files...")
export_dir = os.path.join(SAVE_DIR, 'exported_model')
if os.path.exists(export_dir):
    try:
        for file in os.listdir(export_dir):
            file_path = os.path.join(export_dir, file)
            if os.path.isfile(file_path):
                api.upload_file(
                    path_or_fileobj=file_path,
                    path_in_repo=file,
                    repo_id=REPO_ID,
                    repo_type='model',
                )
                print(f"   ✅ {file}")
    except Exception as e:
        print(f"   ❌ Error: {e}")

# Upload other files
print("\n3. Uploading results and visualizations...")
for filename, description in files:
    filepath = os.path.join(SAVE_DIR, filename)

    if not os.path.exists(filepath):
        print(f"   ⚠️  Skipping {filename} (not found)")
        continue

    try:
        api.upload_file(
            path_or_fileobj=filepath,
            path_in_repo=filename,
            repo_id=REPO_ID,
            repo_type='model',
        )
        size = os.path.getsize(filepath) / (1024 * 1024)
        print(f"   ✅ {filename:30s} ({size:.2f} MB)")
    except Exception as e:
        print(f"   ❌ {filename}: {e}")

print("\n" + "="*70)
print("🎉 UPLOAD COMPLETE!")
print("="*70)
print(f"\n🌐 Your model is now public at:")
print(f"   https://huggingface.co/{REPO_ID}")
print(f"\n📝 View your model card:")
print(f"   https://huggingface.co/{REPO_ID}/blob/main/README.md")
print(f"\n⬇️  Download instructions:")
print(f"   from huggingface_hub import hf_hub_download")
print(f"   model = hf_hub_download('{REPO_ID}', 'best_model.pth')")
````
